## distribution


```python
sns.distplot(tips["total_bill"], kde=False)
#kernel density estimation (KDE) is a non-parametric way to estimate the probability density function of a random variable. In the context of Seaborn's distplot, setting kde=False means that the plot will only display the histogram of the data without overlaying the KDE curve.
```

```python

sns.distplot(tips["total_bill"], kde=True)
```

```python
sns.displot(tips["total_bill"], kde=True,bins=10)
#bins=10 means that the histogram will be divided into 10 equal-width bins to display the distribution of the "total_bill" variable. The kde=True argument indicates that a kernel density estimation curve will be overlaid on the histogram to provide a smoothed estimate of the probability density function of the data.
```

```python
#join plot
#sns.jointplot(x="total_bill",y="tip",data=tips,kind="hex")
#kind="hex" creates a hexagonal binning plot, which is a type of joint plot that uses hexagonal bins to represent the density of data points in the joint distribution of the "total_bill" and "tip" variables. This visualization helps to identify patterns and relationships between the two variables while also showing the distribution of each variable along the axes.

#kind is a parameter in Seaborn's jointplot function that specifies the type of plot to be created for visualizing the relationship between two variables. The available options for kind include "scatter", "reg", "resid", "kde", and "hex". Each option provides a different way to visualize the joint distribution of the data, allowing users to choose the most appropriate plot type based on their specific analysis needs.

```

```python
sns.jointplot(x="total_bill",y="tip",data=tips,kind="hex")
#reg stands for regression. When you set kind="reg" in Seaborn's jointplot function, it creates a scatter plot of the data points along with a regression line that represents the best fit line through the data. This allows you to visualize the relationship between the "total_bill" and "tip" variables while also showing the overall trend in the data. The regression line can help identify whether there is a positive, negative, or no correlation between the two variables.
```

```python
sns.jointplot(x="total_bill",y="tip",data=tips,hue="size") 
#hue="size" adds a color dimension to the joint plot based on the "size" variable in the dataset. This allows you to visualize how the relationship between "total_bill" and "tip" varies across different sizes of groups (e.g., number of people in a party). Each unique value of "size" will be represented by a different color in the plot, making it easier to identify patterns and trends based on group size.
```

```python
#pair plot
sns.pairplot(tips)
#pairplot is a function in Seaborn that creates a grid of scatter plots for each pair of numerical variables in a dataset. It allows you to visualize the relationships between multiple variables simultaneously, making it easier to identify patterns and correlations in the data. Each diagonal plot typically shows the distribution of a single variable, while the off-diagonal plots show the scatter plots of pairs of variables.
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
tips = sns.load_dataset("tips")
print(tips.head())
print("Average Tip:", tips["tip"].mean())
# visualization :bar chart of total bill by day
sns.barplot(x="day", y="total_bill", data=tips)
plt.title("Total Bill by Day")
plt.show()
```

```python
sns.boxplot(x='day', y='total_bill', data=tips)
```

```python
sns.boxplot(x='day', y='total_bill', data=tips,hue="smoker") #split the boxplot by smoker status
```

```python
sns.stripplot(x='day', y='total_bill', data=tips,hue="smoker") #strip plot is used to visualize the distribution of a numerical variable across different categories of a categorical variable. It is similar to a scatter plot but with the points jittered to avoid overlap, making it easier to see the distribution of data points within each category. In Seaborn, you can create a strip plot using the sns.stripplot() function, where you specify the x-axis as the categorical variable and the y-axis as the numerical variable. Strip plots are particularly useful for showing individual data points and identifying any patterns or outliers in the distribution of the numerical variable across different categories.
```

```python
sns.violinplot(x='day', y='total_bill', data=tips) #violin plot is used to visualize the distribution of a numerical variable across different categories of a categorical variable. It combines aspects of a box plot and a kernel density plot, showing both the summary statistics (like median and interquartile range) and the full distribution of the data. In Seaborn, you can create a violin plot using the sns.violinplot() function, where you specify the x-axis as the categorical variable and the y-axis as the numerical variable. Violin plots are particularly useful for comparing the distribution of the numerical variable across different categories while also providing insights into the density of the data points within each category.       
sns.swarmplot(x='day', y='total_bill', data=tips) #swarm plot is used to visualize the distribution of a numerical variable across different categories of a categorical variable. It is similar to a strip plot but with the points arranged in a way that avoids overlap, making it easier to see individual data points and their distribution within each category. In Seaborn, you can create a swarm plot using the sns.swarmplot() function, where you specify the x-axis as the categorical variable and the y-axis as the numerical variable. Swarm plots are particularly useful for showing individual data points and identifying any patterns or outliers in the distribution of the numerical variable across different categories while also providing a clear visualization of the density of data points within each category.
```

```python
#correlation plot is used to visualize the relationship between two numerical variables. It helps to identify patterns, correlations, and trends in the data. In Seaborn, you can create a correlation plot using the sns.heatmap() function, where you specify the correlation matrix of the numerical variables you want to analyze. Correlation plots are particularly useful for showing the strength and direction of the relationship between numerical variables, allowing you to identify which variables are positively or negatively correlated with each other.      

#perfect positive correlation  : +1 (both variables increase together)
#perfect negative correlation :-1 (one increases while the other decreases)
#No correlation : 0 (no relationship between the variables)
```

```python
tc=tips.select_dtypes(include=["float", "int"]).corr()
```

```python
tc=tips.select_dtypes(include='number').corr() # if all are positive you should not remove the columns
```

```python
sns.heatmap(tc, annot=True, cmap="coolwarm") 
#annot=True to display the correlation values on the heatmap, and cmap="coolwarm" to use a color palette that ranges from cool (blue) to warm (red) colors, making it easier to visually distinguish between positive and negative correlations in the data.
```

```python
sns.heatmap(tc, annot=True, cmap="crest") 
```
