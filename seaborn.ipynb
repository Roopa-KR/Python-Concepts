## distribution


```python
sns.distplot(tips["total_bill"], kde=False)
#kernel density estimation (KDE) is a non-parametric way to estimate the probability density function of a random variable. In the context of Seaborn's distplot, setting kde=False means that the plot will only display the histogram of the data without overlaying the KDE curve.
```

```python
sns.displot(tips["total_bill"], kde=True,bins=10)
#bins=10 means that the histogram will be divided into 10 equal-width bins to display the distribution of the "total_bill" variable. The kde=True argument indicates that a kernel density estimation curve will be overlaid on the histogram to provide a smoothed estimate of the probability density function of the data.
```
