#decision tree: A decision tree is a supervised machine learning algorithm that can be used for both classification and regression tasks. It works by recursively splitting the data into subsets based on the values of the input features, creating a tree-like structure where each internal node represents a decision based on a feature, and each leaf node represents a predicted outcome. Decision trees are easy to interpret and visualize, making them popular for various applications. However, they can be prone to overfitting if not properly pruned or regularized.

#decision tree uses entropy and gini index to make decisions. It is a non-parametric model that can be used for both classification and regression tasks. The decision tree algorithm splits the data into subsets based on the feature that provides the best separation of the target variable. The process continues recursively until a stopping criterion is met, such as a maximum depth or minimum number of samples in a leaf node. Decision trees are easy to interpret and can capture complex relationships in the data, but they can also be prone to overfitting if not properly regularized.

How Tree is Built (Step-by-Step)
Start with full dataset
Calculate impurity (Entropy/Gini)
Try splitting on all features
Choose best split (max IG / min Gini)
Divide dataset
Repeat until:
Pure node OR
Max depth reached OR
Minimum samples reached

Overfitting Problem:

Decision Trees easily overfit (learn noise).

->Solution: Pruning
Types:
    Pre-pruning
        Limit depth
        Min samples
    Post-pruning
        Build full tree → cut unnecessary branches

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# Sample data
X = [[2], [4], [6], [8]]
y = [0, 0, 1, 1]

# Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Model
model = DecisionTreeClassifier(criterion='gini', max_depth=2)

# Train
model.fit(X_train, y_train)

# Predict
y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
```
