```python
import pandas as pd
import numpy as np
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"economic_index.csv")
df.head()
```

```python
#drop unnecessary columns
df.drop(columns=['Unnamed: 0','year','month'],axis=1,inplace=True)
```

```python
df.head()
```

```python
#check any columns has null values
df.isnull().sum()
```

```python
sns.pairplot(df ,plot_kws={'alpha':0.7}) #,diag_kind='kde'
```

```python
df.corr()
```

```python
#Check correlation between features
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(),annot=True,cmap='coolwarm',linewidths=0.5)
plt.show()
```

```python
#visualize more closely the relationship between features
sns.scatterplot(x=df['interest_rate'],y=df['unemployment_rate'],color='r')
```

```python
#Regression plot
sns.regplot(x=df['interest_rate'],y=df['index_price'])
```

```python
sns.regplot(x=df['interest_rate'],y=df['unemployment_rate'])
```
