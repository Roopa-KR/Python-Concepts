```python
#Clustering :
#kmeans:
#formula: dist(x, y) = sqrt((x1-y1)^2 + (x2-y2)^2 + ... + (xn-yn)^2)
#kmeans++:
#1. Choose the first center randomly from the data points.
#2. For each data point x, compute D(x), the distance between x and the nearest center that has already been chosen.
#3. Choose the next center from the data points with probability proportional to D(x)^2
#4. Repeat steps 2 and 3 until k centers have been chosen.
#DBSCAN: Density-Based Spatial Clustering of Applications with Noise    
#1. For each point, find the number of points within a given radius (eps).
#2. If the number of points is greater than or equal to a specified minimum (minPts), then the point is a core point and forms a cluster with its neighbors.
#3. If the number of points is less than minPts but greater than 0, then the point is a border point and is assigned to the cluster of its nearest core point.
#4. If the number of points is 0, then the point is considered noise and is not assigned to any cluster.        
#Hierarchical Clustering:
#1. Start with each data point as its own cluster.      
#2. Compute the distance between all pairs of clusters and merge the two closest clusters.
#3. Repeat step 2 until all data points are in a single cluster or until a specified number of clusters is reached. 
#Agglomerative Clustering:
#1. Start with each data point as its own cluster.  
#2. Compute the distance between all pairs of clusters and merge the two closest clusters.
#3. Repeat step 2 until all data points are in a single cluster or until a  specified number of clusters is reached.    
```

```python
#Kmeans:
import numpy as np
def kmeans(X, k, max_iters=100):
    # Step 1: Initialize centroids randomly from the data points
    np.random.seed(0)  # For reproducibility
    centroids = X[np.random.choice(X.shape[0], k, replace=False)]

    for _ in range(max_iters):
        # Step 2: Assign each data point to the nearest centroid
        distances = np.linalg.norm(X[:, np.newaxis] - centroids, axis=2)
        closest_centroids = np.argmin(distances, axis=1)

        # Step 3: Update centroids by calculating the mean of assigned points
        new_centroids = np.array([X[closest_centroids == i].mean(axis=0) for i in range(k)])

        # Check for convergence (if centroids do not change)
        if np.all(centroids == new_centroids):
            break

        centroids = new_centroids

    return centroids, closest_centroids
```

```python
#Kmeans++
import numpy as np
def kmeans_plus_plus(X, k, max_iters=100):
    # Step 1: Choose the first center randomly from the data points
    np.random.seed(0)  # For reproducibility
    centroids = [X[np.random.choice(X.shape[0])]]

    for _ in range(1, k):
        # Step 2: Compute D(x) for each data point
        distances = np.array([min(np.linalg.norm(x - c) ** 2 for c in centroids) for x in X])

        # Step 3: Choose the next center with probability proportional to D(x)^2
        probabilities = distances / distances.sum()
        next_centroid = X[np.random.choice(X.shape[0], p=probabilities)]
        centroids.append(next_centroid)

    centroids = np.array(centroids)

    for _ in range(max_iters):
        # Step 4: Assign each data point to the nearest centroid
        distances = np.linalg.norm(X[:, np.newaxis] - centroids, axis=2)
        closest_centroids = np.argmin(distances, axis=1)

        # Update centroids by calculating the mean of assigned points
        new_centroids = np.array([X[closest_centroids == i].mean(axis=0) for i in range(k)])

        # Check for convergence (if centroids do not change)
        if np.all(centroids == new_centroids):
            break

        centroids = new_centroids

    return centroids, closest_centroids
```
