## What is Pandas?

It has functions for analyzing, cleaning, exploring, and manipulating data.

## Why Use Pandas?
Pandas allows us to analyze big data and make conclusions based on statistical theories.

Pandas can clean messy data sets, and make them readable and relevant.

Relevant data is very important in data science.

#Difference between axis=0 and axis=1 in pandas:


#axis=0: This refers to the vertical axis (rows) in a DataFrame. When you perform operations with axis=0, it means you are applying the operation across rows. For example, if you use df.sum(axis=0), it will calculate the sum of each column (i.e., sum across rows).


#axis=1: This refers to the horizontal axis (columns) in a DataFrame. When you perform operations with axis=1, it means you are applying the operation across columns. For example, if you use df.sum(axis=1), it will calculate the sum of each row (i.e., sum across columns).   

```python
import pandas as pd

df = pd.DataFrame({
    'A': [1, 2, 3],
    'B': [4, 5, 6]
})

print(df)
```

```python
df.sum(axis=0)
```

```python
df.sum(axis=1)
```

```python
#Joining / combining
import pandas as pd

df1 = pd.DataFrame({
    'id': [1, 2, 3],
    'marks': [80, 90, 70]
})

df2 = pd.DataFrame({
    'id': [2, 3, 4],
    'age': [21, 22, 23]
})
```

```python
print(pd.concat([df1, df2], axis=0, ignore_index=True))  # vertical concat
```

```python
print(pd.concat([df1, df2], axis=1, ignore_index=True))  # horizontal concat
```

```python
# Append (same as concat)
print(df1.append(df2))
```

```python
# Merge (like SQL JOIN)
print(pd.merge(df1, df2, on='id', how='outer')) # right join (keep all rows from df2),left join (keep all rows from df1), inner join (keep only matched rows), outer join (keep all rows from both dfs)
#inner join (keep only matched rows) , outer join (keep all rows from both dfs) are also possible
```

```python
# Join (index-based) this is for index based joining, so we need to set the index first
# df1.set_index('id', inplace=True) #inplace=True means modify the original DataFrame, not return a new one
# df2.set_index('id', inplace=True)

print(df1.join(df2,on='id' ,how='outer'))
```

```python
# Combine First
print(df1.combine_first(df2))
#Fill missing values from other dataframe
```
