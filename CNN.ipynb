```python
#Difference between all the activation functions:
#ReLU: Rectified Linear Unit, f(x) = max(0, x),
#Sigmoid: f(x) = 1 / (1 + e^(-x)),
#Softmax: f(x_i) = e^(x_i) / Σ(e^(x_j)) for all j
#tanH: f(x) = (e^(x) - e^(-x)) / (e^(x) + e^(-x))
#The main difference between these activation functions is that ReLU is used for hidden layers, sigmoid is used for binary classification, softmax is used for multi-class classification, and tanH is used for hidden layers when the data is centered around zero.    
```

```python
#difference between sigmoid and softmax:
#Sigmoid is used for binary classification problems, where the output is a single value between 0 and 1, representing the probability of the positive class. Softmax is used for multi-class classification problems, where the output is a vector of values between 0 and 1, representing the probabilities of each class. The softmax function ensures that the sum of the output probabilities is equal to 1, while the sigmoid function does not have this constraint.  
#difference between ReLU and tanH:
#ReLU is a non-linear activation function that outputs the input directly if it is positive, and outputs zero if it is negative. It is computationally efficient and helps to mitigate the vanishing gradient problem. tanH is also a non-linear activation function that outputs values between -1 and 1, which can help to center the data around zero. However, it can suffer from the vanishing gradient problem, especially for large input values. ReLU is generally preferred for hidden layers in deep neural networks, while tanH can be used when the data is centered around zero.   
#difference between ReLU and sigmoid:       
#ReLU is a non-linear activation function that outputs the input directly if it is positive,    and outputs zero if it is negative. It is computationally efficient and helps to mitigate the vanishing gradient problem. Sigmoid is a non-linear activation function that outputs values between 0 and 1, which can be interpreted as probabilities. However, it can suffer from the vanishing gradient problem, especially for large input values. ReLU is generally preferred for hidden layers in deep neural networks, while sigmoid is used for binary classification problems.
#difference between tanH and sigmoid:
#tanH is a non-linear activation function that outputs values between -1 and 1, which can help to center the data around zero. Sigmoid is a non-linear activation function that outputs values between 0 and 1, which can be interpreted as probabilities. The main difference between tanH and sigmoid is that tanH can output negative values, while sigmoid only outputs positive values. This can make tanH more suitable for hidden layers in deep neural networks, while sigmoid is typically used for binary classification problems.    
#difference between ReLU and softmax:
#ReLU is a non-linear activation function that outputs the input directly if it is positive, and outputs zero if it is negative. It is computationally efficient and helps to mitigate the vanishing gradient problem. Softmax is a non-linear activation function that outputs a vector of values between 0 and 1, representing the probabilities of each class. The main difference between ReLU and softmax is that ReLU is used for hidden layers, while softmax is used for the output layer in multi-class classification problems.   
```

```python
#building a cnn model to classify images of cats and dogs using.
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
from tensorflow.keras.preprocessing.image import ImageDataGenerator

```

```python
train_datagen = ImageDataGenerator(
    #Image pixels range: 0–255
    #Neural networks learn better with 0–1
    rescale=1./255,
     #Randomly zooms image in or out by 20%  
     #Why?   
     #Dog may appear big or small in real life.
     #Model learns to recognize dog at any size
    zoom_range=0.2, 
    #Real images are not always straight
    #Helps model handle slanted images
    shear_range=0.2,
    #Dog facing left or right is still a dog
    #Increases dataset size artificially
    horizontal_flip=True
    #THIS WHOLE BLOCK DOES:
    #Normalization
    #Image augmentation
    #Prevents overfitting
)
#this portion will Reads images from folders automatically

train_data = train_datagen.flow_from_directory(
    "dataset/train",
    target_size=(64,64),
    batch_size=4,#Model sees 4 images at a time
    class_mode="binary"
)
```

```python
#Creates a linear stack of layers
#Data flows one layer after another
model = Sequential()

# Conv2D → extracts features from image

# 32 → number of filters (feature detectors)

# (3,3) → filter size (small window)

# relu → removes negative values

# input_shape=(64,64,3)

# 64 × 64 → image size

# 3 → RGB channels
model.add(Conv2D(32, (3,3), activation='relu', input_shape=(64,64,3)))

#Reduces image size by half
#Keeps important features
model.add(MaxPooling2D(2,2))


# Learns more complex features
# Eyes
# Ears
# Face shape
model.add(Conv2D(64, (3,3), activation='relu'))
model.add(MaxPooling2D(2,2))

#Converts 2D feature maps into 1D vector
model.add(Flatten())


# Acts like brain decision layer
# 128 neurons
# Learns final combinations
# ReLU → faster learning
model.add(Dense(128, activation='relu'))
model.add(Dense(1, activation='sigmoid'))


#ReLU is for learning features, Sigmoid is for making the final decision.


# ReLU layers → chefs preparing ingredients
# Sigmoid layer → judge saying Yes / No
```

| Layer type           | Activation | Why                 |
| -------------------- | ---------- | ------------------- |
| Hidden               | ReLU       | Learn features fast |
| Output (Binary)      | Sigmoid    | Probability output  |
| Output (Multi-class) | Softmax    | Class probabilities |


```python
model.compile(
    optimizer='adam',#Optimizer decides how model updates weights.
    loss='binary_crossentropy',#Loss function measures how wrong the prediction is.
    metrics=['accuracy'] #Metric is for human understanding, not learning.
)
#Compile tells the model how to learn and how to judge itself.
```
