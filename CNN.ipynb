```python
#Difference between all the activation functions:
#ReLU: Rectified Linear Unit, f(x) = max(0, x),
#Sigmoid: f(x) = 1 / (1 + e^(-x)),
#Softmax: f(x_i) = e^(x_i) / Σ(e^(x_j)) for all j
#tanH: f(x) = (e^(x) - e^(-x)) / (e^(x) + e^(-x))
#The main difference between these activation functions is that ReLU is used for hidden layers, sigmoid is used for binary classification, softmax is used for multi-class classification, and tanH is used for hidden layers when the data is centered around zero.    
```
